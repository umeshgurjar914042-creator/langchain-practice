```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

टेक्स्ट = """Generative AI models can write text, code and images.

Retrieval-Augmented Generation, or RAG, gives a model access to your own documents. The documents are split into chunks. Each chunk is converted into an embedding and stored in a vector database.

When a user asks a question, the system finds the closest chunks and sends them to the model along with the question."""

splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=40)
टुकड़े = splitter.split_text(टेक्स्ट)

print("टुकड़े:", len(टुकड़े))
for i, t in enumerate(टुकड़े):
    print(i, len(t), "|", t[:60])
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    """नीचे दिए संदर्भ के आधार पर ही जवाब दो।
अगर संदर्भ में जवाब न हो तो कहो "मुझे नहीं पता"।

संदर्भ: {context}

सवाल: {question}"""
)

print(prompt.format(context="वेतन 5 तारीख को आता है।", question="वेतन कब आता है?"))
```
